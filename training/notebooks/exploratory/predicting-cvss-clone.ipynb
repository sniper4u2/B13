```python
import os, json, re, html
import pandas as pd
import numpy as np
from datasets import Dataset
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, f1_score, mean_absolute_error, mean_squared_error
import torch
import torch.nn.functional as F
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    DataCollatorWithPadding,
    AutoModel,
    get_linear_schedule_with_warmup
)

# Disable external logging (W&B etc.)
os.environ["WANDB_DISABLED"] = "true"

# Config
MODEL_NAME = "bert-base-uncased"
MAX_LEN = 256
BATCH_SIZE = 8     # reduce if you hit OOM
EPOCHS = 2         # increase later if you have more time
OUTPUT_DIR = "/kaggle/working"
```

```python
# path to CSV inside Kaggle dataset input
csv_path = "/kaggle/input/csv-reduced/cve_reduced.csv"  

df = pd.read_csv(csv_path)
print("Rows, cols:", df.shape)
df.head()
```

```python
# Adjust column names as needed
TEXT_COL = "summary"        # change to your text column name
SCORE_COL = "cvss"          # change to your numeric cvss base score column
DATE_COL = "pub_date"       # change if you have date; otherwise do random stratified split

# Cleaning function
RE_CLEAN = re.compile(r"\s+|\n|\r")
def clean_text(s):
    if not isinstance(s, str): return ""
    s = html.unescape(s)
    s = RE_CLEAN.sub(" ", s).strip()
    return s

# apply cleaning
df[TEXT_COL] = df[TEXT_COL].astype(str).apply(clean_text)
df[SCORE_COL] = pd.to_numeric(df[SCORE_COL], errors="coerce")
df = df.dropna(subset=[TEXT_COL, SCORE_COL]).reset_index(drop=True)

# map score -> severity (cvss v3.1 buckets)
def score_to_severity(score):
    if score == 0.0:
        return "None"
    if score <= 3.9:
        return "Low"
    if score <= 6.9:
        return "Medium"
    if score <= 8.9:
        return "High"
    return "Critical"

df["severity"] = df[SCORE_COL].apply(score_to_severity)

# Time-based split if you have dates, else stratified
if DATE_COL in df.columns:
    df[DATE_COL] = pd.to_datetime(df[DATE_COL], errors="coerce")
    df = df.sort_values(DATE_COL).reset_index(drop=True)
    n = len(df)
    train_df = df.iloc[:int(0.8*n)].reset_index(drop=True)
    val_df = df.iloc[int(0.8*n):int(0.9*n)].reset_index(drop=True)
    test_df = df.iloc[int(0.9*n):].reset_index(drop=True)
else:
    # stratified split by severity
    from sklearn.model_selection import train_test_split
    train_df, temp_df = train_test_split(df, test_size=0.2, stratify=df["severity"], random_state=42)
    val_df, test_df = train_test_split(temp_df, test_size=0.5, stratify=temp_df["severity"], random_state=42)

print("Train/Val/Test sizes:", len(train_df), len(val_df), len(test_df))

# Label encoding for classification
le = LabelEncoder()
le.fit(train_df["severity"])
train_df["severity_label"] = le.transform(train_df["severity"])
val_df["severity_label"] = le.transform(val_df["severity"])
test_df["severity_label"] = le.transform(test_df["severity"])

# Convert to HF Dataset
hf_train = Dataset.from_pandas(train_df)
hf_val = Dataset.from_pandas(val_df)
hf_test = Dataset.from_pandas(test_df)
```

```python
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer, DataCollatorWithPadding
import numpy as np

# ----- Classification -----
num_labels = len(le.classes_)
cls_model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=num_labels)

def compute_metrics_cls(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=1)
    return {
        "accuracy": accuracy_score(labels, preds),
        "f1": f1_score(labels, preds, average="weighted", zero_division=0)
    }

training_args_cls = TrainingArguments(
    output_dir=os.path.join(OUTPUT_DIR, "cls_results"),
    evaluation_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE,
    num_train_epochs=EPOCHS,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    fp16=True,
    report_to="none",
    remove_unused_columns=False,
)

trainer_cls = Trainer(
    model=cls_model,
    args=training_args_cls,
    train_dataset=hf_train,
    eval_dataset=hf_val,
    tokenizer=tokenizer,
    data_collator=DataCollatorWithPadding(tokenizer),
    compute_metrics=compute_metrics_cls
)

print("== TRAINING CLASSIFICATION ==")
trainer_cls.train()
trainer_cls.save_model(os.path.join(OUTPUT_DIR, "cvss_classification_model"))

eval_results = trainer_cls.evaluate()
print("Validation Results:", eval_results)
```

```python
# ============================================================
# BERT Regression on CVE Dataset (Kaggle-ready)
# ============================================================

!pip install -q transformers datasets accelerate scikit-learn

import pandas as pd
import numpy as np
import torch
import re
import html
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding
)
from datasets import Dataset
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ============================================================
# Configuration
# ============================================================
MODEL_NAME = "bert-base-uncased"
MAX_LEN = 256
BATCH_SIZE = 8
EPOCHS = 3
LR = 2e-5
DATA_PATH = "/kaggle/input/csv-reduced/cve_reduced.csv"  # ✅ adjust to Kaggle input path

# ============================================================
# Load and Prepare Data
# ============================================================
df = pd.read_csv(DATA_PATH)
print(f"Loaded dataset with {len(df)} rows and columns: {df.columns.tolist()}")

# Ensure required columns exist
required_cols = ["summary", "pub_date", "cvss"]
missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# Sort by publication date for time-based split
df = df.sort_values("pub_date").reset_index(drop=True)
n = len(df)
train_end = int(n * 0.8)
val_end = int(n * 0.9)
train_df = df.iloc[:train_end].copy()
val_df = df.iloc[train_end:val_end].copy()
test_df = df.iloc[val_end:].copy()

print(f"Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")

# ============================================================
# Hugging Face Dataset Conversion
# ============================================================
hf_train = Dataset.from_pandas(train_df)
hf_val = Dataset.from_pandas(val_df)

# ============================================================
# Tokenization and Preprocessing
# ============================================================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

RE_CLEAN = re.compile(r"\s+|\\n|\\r")

def clean_text(s):
    if not isinstance(s, str):
        return ""
    s = html.unescape(s)
    s = RE_CLEAN.sub(" ", s)
    return s.strip()

def preprocess_function(batch):
    texts = [clean_text(x) for x in batch["summary"]]
    return tokenizer(texts, truncation=True, padding="max_length", max_length=MAX_LEN)

hf_train = hf_train.map(preprocess_function, batched=True, remove_columns=["summary"])
hf_val = hf_val.map(preprocess_function, batched=True, remove_columns=["summary"])

# Add the regression labels
hf_train = hf_train.add_column("labels", train_df["cvss"].tolist())
hf_val = hf_val.add_column("labels", val_df["cvss"].tolist())

hf_train.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])
hf_val.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])

# ============================================================
# Model Setup
# ============================================================
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=1)
model.config.problem_type = "regression"

# ============================================================
# Training Configuration
# ============================================================
training_args = TrainingArguments(
    output_dir="./reg_results",
    evaluation_strategy="epoch",
    save_strategy="epoch",
    learning_rate=LR,
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE,
    num_train_epochs=EPOCHS,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="mae",
    fp16=torch.cuda.is_available(),
    remove_unused_columns=False,
    logging_dir="./logs",
    logging_strategy="steps",
    logging_steps=100,
)

# ============================================================
# Metrics
# ============================================================
def compute_metrics(eval_pred):
    preds, labels = eval_pred
    if isinstance(preds, tuple):
        preds = preds[0]
    preds = preds.squeeze()
    preds = np.clip(preds, 0.0, 10.0)
    mae = mean_absolute_error(labels, preds)
    mse = mean_squared_error(labels, preds)
    rmse = np.sqrt(mse)
    return {"mae": mae, "rmse": rmse}

# ============================================================
# Trainer
# ============================================================
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=hf_train,
    eval_dataset=hf_val,
    compute_metrics=compute_metrics,
    data_collator=DataCollatorWithPadding(tokenizer),
)

# ============================================================
# Train
# ============================================================
trainer.train()
trainer.save_model(os.path.join(OUTPUT_DIR, "cvss_regression_model"))
# ============================================================
# Evaluate
# ============================================================
eval_results = trainer.evaluate()
print("Final Evaluation Results:", eval_results)
```
